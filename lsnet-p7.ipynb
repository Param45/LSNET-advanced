## Verify Dataset Loading (40% fraction)

Proposal 7 uses only 40% of the ImageNet-1k training data.
The `--dataset-fraction 0.4` flag is passed to `main.py` which
passes it to `build_dataset`, which randomly subsamples 40% of
the training files before building the dataset object.

```python
from data.datasets import build_dataset
```

```python
from types import SimpleNamespace
import time

args = SimpleNamespace(
    data_set="IMNET",
    data_path="/kaggle/input/datasets/sautkin",
    input_size=224,
    color_jitter=0.4,
    aa='rand-m9-mstd0.5-inc1',
    train_interpolation='bicubic',
    reprob=0.25,
    remode='pixel',
    recount=1,
    finetune='',
    inat_category='name',
    dataset_fraction=0.4
)

start = time.time()

dataset_train, n_classes = build_dataset(
    is_train=True,
    args=args
)

print("Classes:", n_classes)
print("Samples:", len(dataset_train))
print("Time:", time.time() - start)
```

```python
start = time.time()

dataset_val, n_classes = build_dataset(
    is_train=False,
    args=args
)

print("Classes:", n_classes)
print("Samples:", len(dataset_val))
print("Time:", time.time() - start)
```

## Run Fine-Tuning — Proposal 7 (Per-Group Micro-SE, 40% Dataset)

**What this does:**
- Loads the pretrained `lsnet_t.pth` weights (`strict=False` — only the new `GroupSE` keys are missing; all backbone keys load correctly).
- Inserts a lightweight per-group Squeeze-and-Excite block inside each `LSConv` layer, applied between the SKA aggregation output and the final `BatchNorm2d`.
- `GroupSE` fc2 weights are zero-initialised and fc2 bias is 1.0, so the gate starts near identity.
- Trains for 5 epochs with 2 warmup epochs on 40% of ImageNet-1k.

```python
# Fine-tune for 5 epochs for Proposal 7 with 40% dataset

%cd /kaggle/working/LSNET-advanced
!python kaggle_run.py --action train_clf --model lsnet_t --finetune /kaggle/working/LSNET-advanced/pretrain/lsnet_t.pth --epochs 14 --warmup-epochs 2 --lr 5e-5 --group-se --dataset-fraction 0.4
```
